# Step 4: Model Training with Imbalanced Data Handling

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

from imblearn.over_sampling import SMOTE
```

```python
df = pd.read_csv("/content/fraud_processed.csv")

X = df.drop("Class", axis=1)
y = df["Class"]
```

```python
non_nan_indices = y.dropna().index
X_filtered = X.loc[non_nan_indices]
y_filtered = y.loc[non_nan_indices]

X_train, X_test, y_train, y_test = train_test_split(
    X_filtered, y_filtered,
    test_size=0.2,
    random_state=42,
    stratify=y_filtered
)
```

```python
smote = SMOTE(random_state=42)

X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

print("Before:", y_train.value_counts())
print("After:", y_train_res.value_counts())
```

```python
log_model = LogisticRegression(max_iter=1000)

log_model.fit(X_train_res, y_train_res)

log_preds = log_model.predict(X_test)

print("Logistic Regression Results:\n")
print(classification_report(y_test, log_preds))
```

```python
rf_model = RandomForestClassifier()

rf_model.fit(X_train_res, y_train_res)

rf_preds = rf_model.predict(X_test)

print("Random Forest Results:\n")
print(classification_report(y_test, rf_preds))
```
